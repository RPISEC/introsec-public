<h2 style="font-size:20px; font-family:Monaco">1.1 - What is Web?</h2>

Welcome to Web! Web, short for Website Exploitation, is a variety of techniques one can use to attack various websites, databases and other online platforms. If you're accessing it through the browser or a network, it's likely Web or related to it.

Web vulnerabilites come in a variety of shapes and sizes. Most boil down to, at their core, giving some sort of malicous user input. A variety of well known vulnerabilty types / classes, in no paticular order, include:

- Path Traversal (Also known as Directory Traversal)
- SSRF (Server Side Request Forgery)
- File upload vulnerabilities
- SQL, NoSQL, OS Injections (Each one is different)
- XSS (Cross Site Scripting)
- SSTI (Server Side Template Injection)
- Authorization / Authentication Misconfigurations
- Web LLM attacks
- Many, many more

OWASP Top 10 keeps an updated list of their most common Web Application Vulnerabilties, of which you can view the most recent list here: https://owasp.org/Top10/2025/. There are tons of other sites which also have good information, feel free to look around for those as well.

<h2 style="font-size:20px; font-family:Monaco">3.3 - Different Common File Types</h2>

Within websites, there's a few types of files that you'll see quite often in the inspect view. These are:

- HTML (HyperText Markup Language)
  - .html or .htm file endings.
  - The structure of the page - the text, images and so on in the order they're shown.
  - Great tutorial: [this](https://www.w3schools.com/html/html_intro.asp).
  - Very relevant for understanding what's going on in a page.
- CSS (Cascading Style Sheets)
  - .css file ending.
  - The styling of the page, so colors, fonts etc. Acts like a static database the HTML pulls from.
  - Whenever you see an "id" or "class" attribute in an HTML file, it's linking that element to a set of CSS rules.
  - Really good tutorial: [this](https://www.w3schools.com/css/css_intro.asp).
  - Not super relevant for CTFs, but is relevant for general website know-how.
- JavaScript
  - .js file ending.
  - The only truely "code" behind the website that you normally can see. If you can see the JavaScript via inspect, you are very likely looking at client-side code; code that runs on your machine, not the server.
  - Tutorial: [this](https://www.w3schools.com/js/default.asp). That being said, whilst HTML and CSS are very unique and you should likely spend some time learning them, JavaScript is similar enough to other programming languages you can understand most of it without spending much time on it.
  - Very relevant for understanding the specifics of what code is happening on any specific web page.

<h2 style="font-size:20px; font-family:Monaco">3.6 - Robots.txt</h2>

Robots.txt is a text file that many websites put in their root to tell web crawlers, bots and the like what they're not allowed to visit. The example one from Google looks like this:

``` text
User-agent: Googlebot
Disallow: /nogooglebot/

User-agent: *
Allow: /

Sitemap: https://www.example.com/sitemap.xml
```

What that actually means to us is not that important, although if you're curious [read about it here](https://developers.google.com/crawling/docs/robots-txt/create-robots-txt). What matters, in this example, is this line:

``` text
Disallow: /nogooglebot/
```

This means that there's a page/folder at /nogooglebot/!! Pages hidden to bots via robots.txt are not always hidden to humans, and since robots.txt is always visible, checking robots.txt can be an easy way to find some hidden pages in a site.
