<h1 style="font-size:30px; font-family:Monaco">1 - Heap Basics</h1>

The stack is a contiguous layout of memory that moves linearly up and down. The heap, on the other hand, is more like a blob of memory that you cut chunks out of when you need them, and put them back when you are done.

<hr style="border:1px solid white">

<h2 style="font-size:20px; font-family:Monaco">1.1 - How is it different?</h2>

Instead of one line of addresses that go from something like 0x5555555560 to 0x7ffffffff0, the heap is a range of addresses that the program can use to create data structures of varying size. The C function `malloc()` is the main perpetrator of this action. The way it does this is by marking a currently marked "free" region of memory that fits the requested size as "in-use" and returns a pointer to the start of that chunk of memory.

Bear in mind, the data stored there is still the same!

It's still just bytes! The heap may not be contiguous, but data can still be written to/read from/executed in (if you are really lucky) there. Keep that in mind as you work on the example.

<hr style="border:1px solid white">

<h2 style="font-size:20px; font-family:Monaco">1.2 - There's a lot more to it!</h2>

The heap is disgustingly complex, and the easiest way to be exposed to this is looking up "heap bin exploitation". That will give you more than enough to chew on over the next 3 years of your exploitation career. Tonight, however, we will be solely focusing on Use-After-Frees.

<h1 style="font-size:30px; font-family:Monaco">2 - Baby's first heap exploit</h1>

While there are a ton of heap exploits (feel free to look up "house heap exploits" for an <em>extensive</em> list), this lecture will only cover a very basic one.

<hr style="border:1px solid white">

<h2 style="font-size:20px; font-family:Monaco">2.1 - Use-After-Free (UAF)</h2>

We talked about how data is allocated on the heap, but what about cleaning it up? On the stack, you just move `RSP` back up and call it a day. On the heap, you use `free()`.

`free()` is a C function that will mark a region of memory previously allocated as unused, and place it in a bin (but it won't clear the data in there!). We aren't going to talk about the bins right now, only one quirk of `malloc()`.

See, when you `malloc()` a chunk, then `free()` it, then `malloc()` another chunk of the same size, it would be silly to go and allocate a brand-new chunk of memory when the one you just used is suitable (temporal caching!). As such, `malloc()` will re-allocate the block of memory that you had previously used.

This is a problem since the data was not cleared before! Data can overlap, leading to some serious corruption! If you have two structs of the same size but different parts are treated as different data types (e.g. `struct1[0]` is a string whereas `struct2[0]` is a pointer), and you try and dereference `struct1[0]`'s string, you might get some weird behavior!
